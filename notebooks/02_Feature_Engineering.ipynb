# Feature Engineering
Create new features and transform existing ones for better model performance.

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

# Load data
train_df = pd.read_csv('../data/sample_train.csv')
print(train_df.head())
```

```python
# Create interaction features
train_df['feature_interaction'] = train_df['feature1'] * train_df['feature2']
train_df['feature_sum'] = train_df['feature1'] + train_df['feature2'] + train_df['feature3']
train_df['feature_ratio'] = train_df['feature1'] / (train_df['feature2'] + 1e-8)

print(train_df.head())
```

```python
# Normalize features
scaler = StandardScaler()
feature_cols = ['feature1', 'feature2', 'feature3']
train_df[feature_cols] = scaler.fit_transform(train_df[feature_cols])

print(train_df.head())
```
