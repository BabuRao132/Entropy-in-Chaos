# Model Evaluation
Evaluate final model performance and analyze results.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix, classification_report, roc_curve, auc
import pickle

# Load the trained model
with open('../outputs/model.pkl', 'rb') as f:
    model = pickle.load(f)

# Load metrics
with open('../outputs/metrics.txt', 'r') as f:
    print(f.read())
```

```python
# Load predictions
submission = pd.read_csv('../outputs/submission.csv')
print(submission.head())
print(f"Total predictions: {len(submission)}")
```

```python
# Feature importance
import pandas as pd
train_df = pd.read_csv('../data/sample_train.csv')
feature_names = train_df.drop(['id', 'target'], axis=1).columns

importance_df = pd.DataFrame({
    'feature': feature_names,
    'importance': model.feature_importances_
}).sort_values('importance', ascending=False)

print(importance_df)
plt.barh(importance_df['feature'], importance_df['importance'])
plt.xlabel('Importance')
plt.title('Feature Importance')
plt.show()
```
